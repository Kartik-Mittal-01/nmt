```python
import re
import sys
```

```python
'''So the whole function simply cleans the decoded text before computing BLEU/chrF/TER, 
   ensuring that tokenization artifacts don't hurt the evaluation metrics.
'''
def clean_text(line: str) -> str:
    cleaned = re.sub(r'\s+([?.!,।॥])', r'\1', line)
    cleaned = cleaned.replace("@@ ", "") 
    cleaned = cleaned.replace(" ##", "")
    return cleaned
```

```python
'''Read each line from the input file, clean tokenization artifacts, and write the cleaned text to the output file'''
def detokenize_file(input_file, output_file):

    with open(input_file, 'r', encoding='utf-8') as f:
        lines = f.readlines()

    with open(output_file, 'w', encoding='utf-8') as f:
        for line in lines:
            f.write(clean_text(line))
```

```python
if __name__ == "__main__":
    input_file = "verify_text_file.txt"
    output_file = "verify_text_file_clean.txt"
    detokenize_file(input_file, output_file)
    print(f"Cleaned {input_file} -> saved to {output_file}")
```
